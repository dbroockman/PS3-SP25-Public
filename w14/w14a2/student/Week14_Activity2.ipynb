In today's assignment, we'll examine a paper by Willer (2009).

Willer theorizes that one reason we see people engage in forms of collective action despite the collective action problem is that groups reward people who contribute to the group with high social status. In turn, being higher status is valuable to individuals because it leads others to evaluate them more highly, defer to their preferences, give them more resources, etc. In this way, participating in collective action might sometimes be in individuals' self-interest, because individuals expect other benefits from their group members down the line if they participate. E.g., if you recycle, people will think you're a good person and be nice; and so, you do recycle, because it's actually in your own interest to do so.

To test this, Willer (2009) conducted a few experiments. The first he calls the "Public Goods Game." This experiment works as follows (quoting and paraphrasing from Willer (2009), and focusing on a few key parts):
1. Groups of six participants in the study came to a university computer lab and were given \\$5.
2. The six people were sat at computers and asked how much of their \\$5 they wanted to contribute to the "public good." They were also told that however much money all six in total gave to the public good would be doubled, and then evenly divided between all participants. Thus, if everyone contributed their entire \\$5 to the public good, then everyone would receive \\$10 (because 6 people * \\$5 = \\$30 contributed in total to the public good, \\$30 * 2 = \\$60 after being doubled, \\$60 / 6 = \\$10 after being divided across all six people). But if no one contributed anything, everyone would leave with only their initial \\$5. This is thus an example of a collective action problem: the group is better off if everyone "does their part" by contributing to the group, but each individual is better off if they don't contribute anything.
3. After playing the game, participants were told how much other participants contributed. E.g., Person 1 would see a list on their computer screen that told them how much Persons 2, 3, 4, 5 and 6 in the game had contribute to the "public good."
4. Phase two of the game then started. Participants were then *randomly assigned* to be paired with another person. In the "High Contribution" condition, they were paired with the person who gave the second most to the "public good." In the "Low Contribution" condition, they were paired with the person who gave the second _least_ to the "public good."
5. The first outcome measure asked participants to rate the other person's motivations on a scale from 0-100, using terms like "honorable," "respected," "prestigious." The first outcome is where on the 0-100 scale participants place the other person.
6. The second outcome measure was measured using a game called the Fear of Greed game. In this game, participants were placed into an economic game where they had to predict whether the other person would act in a greedy manner or not. If they thought the other person would be greedy, they should not cooperate. The outcome is whether the participant chose to cooperate in this Fear of Greed game.
